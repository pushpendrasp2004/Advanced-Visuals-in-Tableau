Question 1 : Create a Heatmap to analyze Treatment Cost by Department and Region.
- Steps:

Drag Department → Rows
Drag Region → Columns
Drag Treatment Cost → Color
Change Marks type to Square
Increase size for better visibility

Result/Insight:

Darker color = higher treatment cost
Helps identify which department-region combination is most expensive


Question 2 : Create a Highlight Table showing Satisfaction Score by Department.
- Steps:

Drag Department → Rows
Drag Satisfaction Score → Text
Again drag Satisfaction Score → Color
Choose Highlight Table from Show Me

Result:

Displays satisfaction with color intensity
Quickly identifies high and low performing departments


Question 3 : Create a Bump Chart to show ranking of Departments based on Treatment Cost over months.
- Steps:

Drag Month (Order Date) → Columns
Drag Treatment Cost → Rows
Drag Department → Color
Create Rank Calculation:
Right-click Treatment Cost → Quick Table Calculation → Rank
Reverse axis (Rank 1 at top)

Result:

Shows how department ranking changes over time
Useful to track performance trends

Question 4 : Create a Butterfly Chart comparing Male vs Female patient count by Department.
- Steps:

Drag Department → Rows
Drag Patient Count → Columns (twice)
Filter one for Male, other for Female
Make one axis negative (calculated field):

Male Count = -SUM([Patient Count])

Synchronize axes

Result:

Left side = Male, Right side = Female
Easy gender comparison across departments


Question 5 :  Create a Dual Axis Chart to compare Average Age and Treatment Cost by Department.
- Steps:

Drag Department → Columns
Drag Average Age → Rows
Drag Treatment Cost → Rows
Right-click second axis → Dual Axis
Synchronize axis
Use different colors

Result:

Compares age group vs cost
Helps understand if older patients incur higher costs

Question 6 : Create a Sunburst Chart showing Department → Diagnosis hierarchy by Treatment Cost.
- Steps:

Create hierarchy:
Department → Diagnosis
Use Pie chart
Add fields:
Level 1 (inner): Department
Level 2 (outer): Diagnosis
Drag Treatment Cost → Angle

Result:

Shows contribution of diagnosis within departments
Helps identify cost-heavy treatments
